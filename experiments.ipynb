```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,LabelEncoder
import pickle
```

```python
## Load the dataset
data=pd.read_csv('Churn_Modelling.csv')
data.head()
```

```python
## Preprocess Data
## Drop unnecessary columns
data.drop(['RowNumber','CustomerId','Surname'],axis=1)
data
```

```python
## Encoding categorical variables
label_encoder_gender = LabelEncoder()
data['Gender'] = label_encoder_gender.fit_transform(data['Gender'])
data
```

```python
## Do preprocessing on column Geography using OneHotEncoding
from sklearn.preprocessing import OneHotEncoder
onehot_encoder_geo = OneHotEncoder()  # drop='
geo_encoder=onehot_encoder_geo.fit_transform(data[['Geography']]).toarray()
geo_encoder
```

```python
onehot_encoder_geo.get_feature_names_out(['Geography'])
```

```python
geo_encoder_df = pd.DataFrame(geo_encoder, columns=onehot_encoder_geo.get_feature_names_out(['Geography']))
geo_encoder_df
```
